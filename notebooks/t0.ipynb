```python

from pydantic import BaseModel

class Qux(BaseModel):
    quxint: int
    quxstr: str

class Baz(BaseModel):
    bazint: int
    bazstr: str
    bazqux: Qux

class Bar(BaseModel):
    barint: int
    barstr: str

class Foo(BaseModel):
    fooint: int
    foostr: str
    foobar: Bar
    foobaz: Baz
```

```python
Foo.model_validate({
    "fooint": 0,
    "foostr": "0",
    "foobar": {
        "barint": 1,
        "barstr": "1",
    },
    "foobaz": {
        "bazint": 2,
        "bazstr": "2",
        "bazqux": {
            "quxint": 3,
            "quxstr": "3",
        },
    }
})
```

```python

from pydantic_magic.magic_notation_model import MagicModel, magic_notation

class MagicQux(MagicModel):
    quxint: int
    quxstr: str

class MagicBaz(MagicModel):
    bazint: int
    bazstr: str
    bazqux: MagicQux

class MagicBar(MagicModel):
    barint: int
    barstr: str

class MagicFoo(MagicModel):
    fooint: int
    foostr: str
    foobar: MagicBar
    foobaz: list[MagicBaz]
```

```python
import json
from pydantic_magic.magic_notation_model import MagicModel, magic_notation

p = magic_notation({
    "fooint": 0,
    "foostr": "0",
    "foobar_barint": 1,
    "foobar_barstr": "1",
    "foobaz_0_bazint": 2.0,
    "foobaz_0_bazstr": "2.0",
    "foobaz_1_bazint": 2.1,
    "foobaz_1_bazstr": "2.1",
    "foobaz_2_bazint": 2.2,
    "foobaz_2_bazstr": "2.2",
    # "foobaz_bazqux_quxint": 3,
    # "foobaz_bazqux_quxstr": "3",
})

print(json.dumps(p, indent=4))
```
