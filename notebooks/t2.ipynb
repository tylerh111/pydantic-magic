```python

from pydantic import BaseModel, model_validator

class Test0(BaseModel):
    test0: int = 0

    @model_validator(mode="wrap")
    def vtest0(cls, v, handler):
        print("vtest0")
        self = handler(v)
        print("vtest0 (done)")
        return self


class Test1(Test0):
    test1: int = 1

    @model_validator(mode="before")
    def vtest1(cls, v):
        print("vtest1")
        return v


class Test2(Test1):
    test2: int = 2

    @model_validator(mode="before")
    def vtest2(cls, v):
        print("vtest2")
        return v


class Test3(Test2):
    test3: int = 3

    @model_validator(mode="before")
    def vtest3(cls, v):
        print("vtest3")
        return v

Test3()
```

```python

from pydantic import BaseModel
from pydantic_magic import magic_variant

@magic_variant
class Test(BaseModel):
    pass

class TestAlternate(Test):
    foo: int

print(Test.model_variant)

value = {
    "foo": 0,
}

expected = TestAlternate.model_validate(value)

print(value)

model = Test.model_validate(value)
# model = Test(**value)

assert model == expected
```
